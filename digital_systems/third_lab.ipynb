```python

def roll(h,n):
    h_out = np.roll(h,n)
    print(h_out)
    return h
```

```python
h = np.array([4,5,6])
h_right_1 = roll(h,1)
h_right_2 = roll(h,2)
h_left_1 = roll(h,-1)
h_left_1 = roll(h,-2)
```

```python
#second way to do circular convolution

def circular_convolution(x,h):
    N = len(x)
    y = np.zeros(N)
    for n in range(N):
        for k in range(N):
            y[n] += x[k]*h[(n-k)%N]
    return y

x = np.array([1,2,3])
h = np.array([4,5,6])
y_circular = circular_convolution(x,h)

print("circular convolution:", y_circular)
    
```

```python
#frequency response

delta = lambda n: n==0
a = [1,-1.1,0.5,0,0.3]
b = [0.5,-0.2,0,0,0]
x = np.zeros(10)

for n in range(0,10,1):
    temp = int(delta(n))
    x[n] = temp
n = range(0,10,1)
h = sp.signal.lfilter(b,a,x)
plt.stem(n,h)
```

```python
#frequency response
x = [5,1,1,1,0,0,1,1,1,0]
y = np.convolve(x,h);

#second way
y2 = sp.signal.lfilter(b,a,x)

plt.plot(y)
plt.plot(y2,"r")
plt.show()
```

```python
#frequecy

a = [1,-0.3]
b = [7,0]
w,h = sp.signal.freqz(b,a)
angles = np.unwrap(np.angle(h))

plt.subplot(2,1,1)
plt.plot(w/max(w),20*np.log10(abs(h)))
plt.title("Frequency response")
plt.subplot(2,1,2)
plt.plot(w/max(w),angles*180/np.pi)
plt.title("Phase response")
```
