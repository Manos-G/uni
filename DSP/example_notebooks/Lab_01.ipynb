# Unit Step 

```python
import numpy as np 
import matplotlib.pyplot as plt

n = range(-2, 6, 1)
y = []
for i in range(len(n)):
    temp = (1 if n[i]>=0 else 0)
    y.append(temp)
    
print(n)
print(y)

#plotting the graph
plt.stem(n, y)
plt.axis([-2.1, 5.1, -0.1, 1.2])
plt.show()
```

# Exponential 

```python
a = 0.9
n = range(0, 26, 1)
y = []

for i in range(len(n)):
    temp = a**n[i]
    y.append(temp)
    

#Plotting the graph
plt.stem(y)
plt.axis([-0.3, 25.1, -0.1, 1.1])
plt.xlabel('  n--->  ')
plt.ylabel('Amplitude ----> ')
plt.title('Exponential signal for 0 < a < 1')
plt.grid(True)
plt.show()
```

# διακριτή ακολουθία επιβράδυνσης

```python
def dk(k,min_n,max_n):
    n=range(min_n,max_n,1);
    d=np.zeros(len(n));
    for i in range(1, len(n)):
        if n[i]==k:
            d[i]=1;
            break;
    return n,d
```

```python
[x,y]=dk(10,0,17);
```

```python
print(x)
```

```python
print(y)
```

```python
#plotting the graph
plt.stem(x, y)
plt.show()
```

```python
[n2, x2]=dk(5,-3,8);
x2=x2*0.9;
```

```python
#plotting the graph
plt.stem(n2, x2)
plt.show()
```

```python
[n1,d1]=dk(-3,-6,8);
[n2,d2]=dk(0,-6,8);
[n3,d3]=dk(6,-6,8);
```

```python
plt.stem(n1,d1,'r')
plt.stem(n2,d2,'g')
plt.stem(n3,d3,'b')
```
