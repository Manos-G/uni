Calculate and graph the frequency response ( magnitude and phase) of the following filter. 

H(z) = (0.2 + 0.4z^-1 + 0.6z^-2 + 0.4z^-3 + 0.2z^-4)

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal

# Define the filter coefficients directly
b = np.array([0.2, 0.4, 0.6, 0.4, 0.2])
a = np.array([1.0])

# Compute the frequency response
w, h = signal.freqz(b, a)

# Plot the frequency response
fig, ax = plt.subplots()
ax.plot(w, 20 * np.log10(abs(h)))
ax.set(title='Frequency response', xlabel='Frequency', ylabel='Magnitude (dB)')
ax.grid(True)

plt.show()
```

Calculate and graph the frequency response ( magnitude and phase) of the following filter. 

H(z) = (1 - 0.5z^-1) / (1 + 0.8z^-1)

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal

# Define the filter coefficients
b = np.array([1, -0.5])
a = np.array([1, 0.8])

# Compute the frequency response
w, h = signal.freqz(b, a)

# Compute the phase response and unwrap it
phase = np.unwrap(np.angle(h))

# Plot the phase response
fig, ax = plt.subplots()
ax.plot(w, phase)
ax.set(title='Phase response', xlabel='Frequency', ylabel='Phase (radians)')
ax.grid(True)

plt.show()
```
