```python
import numpy as np
import matplotlib.pyplot as plt
import scipy
from scipy import signal

delta = lambda n: n==0
a = [1, -1.1, 0.5, 0, 0.3]; 
b = [0.5, -0.2, 0, 0, 0]; 
x=np.zeros(10)

for n in range(0,10,1):
    temp=int(delta(n)) 
    x[n]=temp
    
n=range(0,10,1)
h= scipy.signal.lfilter(b,a,x)
plt.stem(n,h)
```

```python
x=[5, 1, 1, 1, 0, 0, 1, 1, 1, 0]; 
y=np.convolve(x, h);
plt.plot(y) 
plt.show()
```

```python
y2=scipy.signal.lfilter(b, a, x); 
plt.plot(y);
plt.plot(y2,'r');
```

```python
a=[1, -0.3]
b=[0.7, 0]
w,h= scipy.signal.freqz(b,a)
angles = np.unwrap(np.angle(h))

plt.subplot(2, 1, 1)
plt.plot(w/max(w), 20 * np.log10(abs(h)))
plt.subplot(2, 1, 2)
plt.plot(w/max(w), angles)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import freqz

# Define the impulse response of the FIR filter
h = np.array([0.25, 0.5, 0.25])

# Compute the frequency response
w, H = freqz(h)

# Plot the magnitude response
plt.figure()
plt.subplot(2, 1, 1)
plt.plot(w, np.abs(H), 'b')
plt.title('Frequency Response')
plt.xlabel('Frequency (rad/sample)')
plt.ylabel('Magnitude')
plt.grid()

# Plot the unwrapped phase response
plt.subplot(2, 1, 2)
plt.plot(w, np.unwrap(np.angle(H)), 'b')
plt.xlabel('Frequency (rad/sample)')
plt.ylabel('Unwrapped Phase (radians)')
plt.grid()

plt.tight_layout()
plt.show()
```
